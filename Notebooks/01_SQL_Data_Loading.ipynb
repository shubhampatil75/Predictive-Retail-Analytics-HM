# Database Setup: H&M Personalized Fashion Recommendations

This notebook will take our raw CSV files (`article.csv`, `customer.csv`, and `transactions.csv`) and ingest them into a serverless SQLite Database file (`hm_database.db`). 

Once this is run, you can use SQL to instantly query millions of rows without overloading your RAM!

```python
import sqlite3
import pandas as pd
import os

# Define exact absolute paths so this runs perfectly no matter what
BASE_DIR = r'D:\Imarticus Learning- PGA 45\GitHub Projects\h & m db'
DB_PATH = os.path.join(BASE_DIR, 'hm_database.db')
DATA_DIR = os.path.join(BASE_DIR, 'data')

print("Connecting to SQLite database (creating it if it doesn't exist)...")
conn = sqlite3.connect(DB_PATH)
```

### 1. Load Customers and Articles
These files are smaller and will load quickly.

```python
print("Loading customer.csv into SQL...")
df_customers = pd.read_csv(os.path.join(DATA_DIR, 'customer.csv'))
df_customers.to_sql('customers', conn, index=False, if_exists='replace')
print("Customers loaded successfully!")

print("Loading article.csv into SQL...")
df_articles = pd.read_csv(os.path.join(DATA_DIR, 'article.csv'))
df_articles.to_sql('articles', conn, index=False, if_exists='replace')
print("Articles loaded successfully!")
```

### 2. Load Transactions (15 Million Rows)
This is a massive file. We will process it in chunks to avoid memory overload. This may take a few minutes.

```python
print("Loading transactions.csv into SQL (Processing in chunks, Please wait...)...")
chunk_size = 500000
for i, chunk in enumerate(pd.read_csv(os.path.join(DATA_DIR, 'transactions.csv'), chunksize=chunk_size)):
    print(f"  Processing chunk {i+1}...")
    if_exists = 'replace' if i == 0 else 'append'
    chunk.to_sql('transactions', conn, if_exists=if_exists, index=False)
    
print("\nTransactions loaded successfully!")
```

### 3. Verify and Close

```python
# Quick check to see what tables were created
tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)
print("\nTables created in the database:")
display(tables)

# Always close the connection when you're done!
conn.close()
print("\nDatabase setup is 100% complete! You can now use the SQL EDA Notebook.")
```
